## 1.2 OVER DE DATASET OP WIKIDATA



Eerst en vooral is het nuttig om even stil te staan bij de werking en structuur van Wikidata. ###**Algemene structuur Wikidata**

Als we op de website van Wikidata 'Guido Gezelle' ingeven in de zoekbalk, is het resultaat een item van Guido Gezelle met een unieke ID-nummer (in dit geval een Q-nummer). Aan de hand van dit item zullen we de algemene opzet van Wikidata toelichten. Door op de link te klikken kan je meevolgen met onderstaande uitleg.

Voorbeeld: 'Guido Gezelle'-record
link: https://www.wikidata.org/wiki/Q336977

**Triples**

De structuur van het datamodel van Wikidata is niet object-georienteerd, maar gebaseerd op 'subjecten'. De data bestaat uit relaties, vastgelegd in zogenaamde 'triples'. Een 'triple' bestaat uit drie delen, een zogenaamde subject-property-object relatie. In de logica wordt ook gesproken over een subject-predicaat stijl.

Voorbeelden van triples: een kat (subject) heeft delen (property) zoals poten (object), oren (object), een snoet (object), enzovoort.
Een toegepast voorbeeld: Guido Gezelle (subject) is een instantie van (property) een mens (object); of Guido Gezelle (subject) heeft het geslacht (property) man (object); of Guido Gezelle (subject 1) heeft een significante persoonsrelatie (property 1) met een persoon (object 1) en deze persoon (subject 2) heeft de rol van (property 2) correspondent (object 2).

Opmerking: het laatste voorbeeld toont de gelaagdheid van dit subject-eigenschap-object systeem. Elke object kan een nieuw subject vormen waaraan vervolgens nieuwe eigenschappen kunnen worden toegekend. Zo is alle informatie in theorie aan elkaar gelinkt via verschillende relaties of eigenschappen.

**Items, statements, properties**

De Wikidata-repository bestaat voornamelijk uit items, elk met een label, een beschrijving en een willekeurig aantal aliassen. Items worden uniek geïdentificeerd door een Q gevolgd door een nummer, zoals Guido Gezelle (Q336977). Andere voorbeelden van items zijn: de stad Brugge (https://www.wikidata.org/wiki/Q12994); KANTL (Koninklijke Academie voor Nederlandse Taal- en Letterkunde; Q1268453), correspondent (Q3589290), mens (Q5), enzovoort.

Beweringen (statements) beschrijven gedetailleerde kenmerken van een item en bestaan ​​uit een eigenschap (property) en een waarde (value). Eigenschappen in Wikidata hebben een 'P' gevolgd door een nummer, zoals met 'heeft geslacht of gender' (P21). Andere voorbeelden zijn: 'is instantie van' (P31), 'heeft significante relatie met' (P3342), 'in de rol van' (P3831), enzovoort.

Met deze kennis is het mogelijk om logische formules te noteren die van belang worden voor het vormen van SPARQL-queries. In plaats van de labels te gebruiken, kan er ook gewerkt worden met de unieke identifiers, in dit geval de 'Q' en 'P'-nummers. Met deze werkwijze kan mensentaal omgezet worden in een logische taal voor machines.

Voorbeeld:

'Guido Gezelle is een instantie van een mens'
kan genoteerd worden als 'Q336977 - P31 - Q5'

##2.2 SPARQL-QUERIES

### Voorbeeldquery 1: Select, Where, Service

**Wikidata Query Service**

In dit deel wordt getoond hoe een SPARQL-querie kan opgebouwd worden met de Wikidata query service. Ik toon hier een eerste querievoorbeeld dat als resultaat alle personen moet geven die gelinkt zijn aan het item Guido Gezelle. Vervolgens licht ik toe hoe dit voorbeeld kan opgebouwd worden.


````python
#SPARQL-querievoorbeeld 1 voor alle personen gelinkt aan Guido Gezelle
SELECT ?person ?personLabel WHERE {
  # Start met personen te zoeken die een relatie hebben tot Guido Gezelle
  wd:Q336977 ?relatie ?person.

  # persoon is een mens
  ?person wdt:P31 wd:Q5.

  # Labels ophalen: probeer Nederlands (nl), val terug op Engels (en)
  SERVICE wikibase:label {
  bd:serviceParam wikibase:language "[AUTO_LANGUAGE],nl,en".
    ?person rdfs:label ?personLabel.
  }
}
````


**SPARQL**

Eerder werd al gekeken naar de werking en de structuur van de data op wikidata. Er werd dieper ingegaan op de 'triple'-structuur van data. Data valt te noteren in termen van items (Q) en eigenschappen (P). Als voorbeeld werd getoond dat 'Guido Gezelle is een mens' geschreven kan worden als Q - P - Q. Op die informatie kan verder gebouwd worden om een querie samen te stellen.

De essentie van een SPARQL-querie komt neer op het schrijven van een lijst variabelen die je wil oproepen en te formuleren aan welke voorwaarden deze variabelen moeten voldoen om ingevuld te raken. De specifieke syntax laat ik hier buiten beschouwing, omdat deze reeds goed gedocumenteerd is. Hieronder zal het principe van parameters en voorwaarden formuleren, getoond worden met een voorbeeld.
https://www.wikidata.org/wiki/Wikidata:SPARQL_tutorial

**1. SELECT**

De query hieronder vormt een SPARQL-query waarmee me willen te weten komen welke personen allemaal aan Guido gelinkt zijn.Daarvoor geven we eerst aan welke parameters we willen krijgen als resultaat, in dit geval de personen als Q-nummers (?person) en de namen van de personen in mensentaal (?personLabel). Deze namen van de parameters zijn vrij naar keuze. '?person' had ook '?linkedEntity' kunnen noemen. Enige voorwaarde is dat een parameter altijd met een vraagteken begint.

**2. WHERE**

Vervolgens bestaat de querie uit een 'where'-clausule ' waar we deze parameters willen invullen aan de hand van enkele voorwaarden. De 'where' leest dus als 'waarvoor geldt'. In het geval van onderstaande query zijn er twee voorwaarden opgesteld. 1) de eerste voorwaarde formuleert dat de persoon is gelinkt aan guido (te lezen als: Guido heeft een willekeurige relatie met ?person). Eigenlijk staat er genoteerd 'Guido Gezelle' (Q) 'heeft een willekeurige eigenschap met' (alle p-nummers) 'een willekeurige entiteit' (alle mogelijke Q) genaamd ?person. Hiermee verkrijgen we dus alle entiteiten (ook niet-personen) die op een of andere manier gelinkt zijn aan Guido Gezelle. Om tot enkel mensen in plaats van alle entiteiten te komen, moet een tweede voorwaarde geformuleerd worden. 2) de tweede voorwaarde zorgt ervoor dat de entiteit een mens is. Dit kan door te noteren dat ?person een eigenschap 'is instantie van' heeft met als waarde/object 'mens'. de persoon is een instantie van een mens (te lezen als: ?person is een mens). De laatste voorwaarde is van belang, omdat zonder die voorwaarde alle gekoppelde entiteiten (ook niet-personen) zullen getoond worden. Bij de querie zelf is ook wat commentaar toegevoegd om de code zelf beter te begrijpen.

**3. SERVICE**

Het laatste deel van de query gaat over het ophalen van de juiste (taalgerelateerde) labels in the wikibase-service (SERVICE). Binnen de wikibase:label worden parameters omtrent taal aangevraagd, in dit geval automatisch nederlands en anders het engelse label. Op deze manier worden niet enkel de Q-nummers, maar ook de labels (mensentaal) opgevraagd.




````python
#SPARQL-querievoorbeeld 1 voor alle personen gelinkt aan Guido Gezelle
SELECT ?person ?personLabel WHERE {
  # Start met personen te zoeken die een relatie hebben tot Guido Gezelle
  wd:Q336977 ?relatie ?person.

  # persoon is een mens
  ?person wdt:P31 wd:Q5.

  # Labels ophalen: probeer Nederlands (nl), val terug op Engels (en)
  SERVICE wikibase:label {
  bd:serviceParam wikibase:language "[AUTO_LANGUAGE],nl,en".
    ?person rdfs:label ?personLabel.
  }
}
````

Opmerking

###Voorbeeldquery 2: prefixen, binden en union

Tot nu toe werd de query vooral conceptueel toegelicht. uiteraard zijn er nog zaken die toegelicht kunnen worden in deze code. De meeste van deze zaken zijn echter syntactische elementen. Met een tweede voorbeeld wil ik andere zaken toelichten en een kleine syntactische complexiteiten toevoegen.

**wd/wdt vs. p,ps en pq**

de verschillende prefixen of afkortingen 'wd, wdt, p,ps en pq die gebruikt worden, bepalen welk deel van een Wikidata-statement je precies opvraagt in de query. In wat volgt worden wd, wdt, p, ps en pq afzonderlijk verder besproken.

1. Directe Eigenschap (wd of wdt): Prefixen wd en wdt worden gebruikt om directe entiteiten op te roepen.  wd wordt gebruikt om een item op te roepen in wikidata, wdt om op directe wijze een property op te roepen.  Het haalt de directe waarde van een eigenschap op zonder de bijbehorende kwalificaties (contextuele informatie).

Voorbeeld: ?guido wdt:P31 wd:Q5. Waarbij p31 verwijst naar de property 'is instantie van' en q5 verwijdt naar de item 'mens'

Leest: "Guido Gezelle (?guido) heeft als instantie van (wdt:P31) de waarde mens (wd:Q5)." Dit is snel en eenvoudig voor feiten zonder context.

2. Gekwalificeerde Statements (p, ps, pq)

Wanneer je een eigenschap met extra context (kwalificaties) wilt doorzoeken – zoals de rol Q3589290 (correspondent) in dit geval – moet je het statement in drie stappen ontleden met behulp van p, ps, en pq. Als we 'wd:Q336977 wdt:P3342 wd: Q3589290 gebruiken in een query zal dit helaas niet werken. Er zijn meer stappen nodig om tot deze data te komen.

**A. Statement-knoop (p)**

Prefix 'p:' verwijst naar de statement-knoop (een tijdelijk, anoniem knooppunt) dat het feit vertegenwoordigt. Dit is de verbinding van het item naar het feit zelf. Een tussenpunt om tot de verdere informatie te komen, zeg maar.

In de query schrijf je: ?guido p:P3342 ?statement.

Leest: "Guido Gezelle (?guido) heeft een statement (p:P3342) over 'relevant persoon' dat we ?statement noemen."

We kunnen nu verder bevragen in het submenu van P3342.

**B. Statement Waarde (ps)**

Prefix 'ps' haalt de hoofdwaarde (de kern van het feit) op van de statement-knoop. In het geval van de data over Guido Gezelle gaat het hier over een persoon als hoofdwaarde.

In de query je vervolgens onder de 'p:'-statement: ?statement ps:P3342 ?person.

Leest: "Van die statement-knoop (?statement) is de hoofdwaarde (ps:P3342) de persoon (?person)."

**C. Statement Kwalificatie (pq)**

Prefix 'pq:' haalt de kwalificaties (context, zoals datum, rol, etc.) op die aan het statement zijn toegevoegd.

In de query schrijven we een derde lijn: ?statement pq:P3831 wd:Q3589290

Leest: "Aan die statement-knoop (?statement) hangt de kwalificatie (pq:P3831, 'heeft de rol van') met de waarde correspondent (?correspondent_role)."

opmerking: Het is mogelijk wat verwarrend om de drie statements bij elkaar te begrijpen. In de praktijk went het snel om deze toe te passen, aangezien het een standaard is in de opbouw van vele SPARQL-queries.

Samengevat:

Wd en wdt voor directe, simpele feiten. geen context nodig (bv. P31 - instantie van). p, ps en pq om complexe feiten op te halen. (in de praktijk ook vaak als wdt of wd niet werkt)P-Statement linkt item aan de statement-knoop.  Altijd nodig bij het doorzoeken van kwalificaties. ps: Statement Waarde    Haalt de kernwaarde van het statement op.   Altijd nodig bij het doorzoeken van kwalificaties. pq: Kwalificatie    Haalt de context (kwalificatie) van het statement op.   Wanneer je contextuele details (zoals de rol) wilt controleren.

Voorbeelden toevoegen: 3 lijnen code zien er dan zo uit.

**binden**

Het eerste dat we doen is het binden van variabelen aan bepaalde Q-nummers. Dat kan met de functie 'BIND'. Zo moeten we de Q-nummers niet blijven herhalen in de code.

Voorbeeld:  BIND(wd:Q336977 AS ?guido)`

in plaats van het Q336977 te gebruiken, kunnen we deze ID hernoemen naar Guido. vanaf dat moment kunnen we overal ?guido schrijven in plaats van het nummer. In de query doen we dat ook voor het item 'correspondent', omdat dat gemakkelijker en sneller noteert. De code wordt hierdoor ook meer leesbaar.

**union**

Aangezien we zowel de personen willen die geschreven hebben naar Guido, als de personen die ontvangen hebben van Guido, moeten we de query in twee richtingen opstellen. Eerder werd al vermeld dat de rol van correspondent niet noodzakelijk symmetrisch is. Dat leidt tot where-statements die bevragen wie gestuurd heeft naar Guido Gezelle én daarnaast where-statements over wie ontvangen heeft van Guido Gezelle.
Om deze twee groepen te verbinden in een zoekresultaat, wordt de functie UNION gebruikt. Op die manier kunnen. Als we UNION niet gebruiken, krijgen we enkel personen die zowel gestuurd hebben naar als ontvangen hebben van Guido Gezelle. Personen die enkel gestuurd of ontvangen hebben, zullen dan niet in de zoekresultaten opduiken.

````python
## SPARQL-queryvoorbeeld 2: Welke personen hebben brieven gestuurd aan of ontvangen van Guido Gezelle?
# persoon stuurt = persoon is correspondent; persoon ontvangt = Guido is correspondent
SELECT DISTINCT ?person ?personLabel WHERE {
  # Bind Guido Gezelle's QID
  BIND(wd:Q336977 AS ?guido)
  BIND(wd:Q3589290 AS ?correspondent) # Correspondent item

  # Zoek personen die als correspondent van Guido vermeld staan (Query 7.1)
  {
    ?guido p:P3342 ?statement.         # Guido heeft een statement over een 'relevant person'
    ?statement ps:P3342 ?person.        # De 'person' is de relevante persoon
    ?statement pq:P3831 ?correspondent. # De kwalificatie van de relevante persoon is 'correspondent'
  }
  UNION
  # Zoek personen bij wie Guido als correspondent vermeld staat (Query 7.2)
  {
    ?person p:P3342 ?statement.         # De 'person' heeft een statement over een 'relevant person'
    ?statement ps:P3342 ?guido.         # Guido is de relevante persoon
    ?statement pq:P3831 ?correspondent. # De kwalificatie van de relevante persoon is 'correspondent'
  }

  # Haal labels op: probeer Nederlands (nl) en val terug op Engels (en)
  SERVICE wikibase:label {
    bd:serviceParam wikibase:language "nl,en".
    ?person rdfs:label ?personLabel.
  }
}
````

met bovenstaande query krijgen we een lijst van personen die ofwel gestuurd ofwel ontvangen hebben, maar de richting wordt hierbij niet meegegeven, omdat de resultaten van elke richting in een gezamenlijke lijst worden geplaatst.
De BIND-functie kan daarom ook handig zijn om labels toe te voegen in verband met de richting (bv gestuurd of ontvangen). In dit geval heb ik daarom deze twee lijnen code toegevoegd aan de query:


```
BIND("Gestuurd door Gezelle" AS ?richting) # Voeg de richting toe
BIND("Ontvangen door Gezelle" AS ?richting) # Voeg de richting toe
```



De query ziet er dan zoals onderstaande uit. De zoekresultaten geven dan een extra kolom waarbij de richting van de correspondentie wordt getoond.

````python
## 🤝 Guido Gezelle: Alle Correspondenten die brieven hebben gestuurd of ontvangen
SELECT DISTINCT ?person ?personLabel ?richting WHERE {
  # Bind Guido Gezelle's QID
  BIND(wd:Q336977 AS ?guido)
  BIND(wd:Q3589290 AS ?correspondent_role) # Correspondent item

  # Zoek personen die als correspondent van Guido vermeld staan (Query 7.1)
  {
    ?guido p:P3342 ?statement.         # Guido heeft een statement over een 'relevant person'
    ?statement ps:P3342 ?person.        # De 'person' is de relevante persoon
    ?statement pq:P3831 ?correspondent_role. # De kwalificatie is 'correspondent'
    BIND("Gestuurd door Gezelle" AS ?richting) # Voeg de richting toe
  }
  UNION
  # Zoek personen bij wie Guido als correspondent vermeld staat (Query 7.2)
  {
    ?person p:P3342 ?statement.         # De 'person' heeft een statement over een 'relevant person'
    ?statement ps:P3342 ?guido.         # Guido is de relevante persoon
    ?statement pq:P3831 ?correspondent_role. # De kwalificatie is 'correspondent'
    BIND("Ontvangen door Gezelle" AS ?richting) # Voeg de richting toe
  }

  # Haal labels op: probeer Nederlands (nl) en val terug op Engels (en)
  SERVICE wikibase:label {
    bd:serviceParam wikibase:language "nl,en".
    ?person rdfs:label ?personLabel.
  }
}
````
